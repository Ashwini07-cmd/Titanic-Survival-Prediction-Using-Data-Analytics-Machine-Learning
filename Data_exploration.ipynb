```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("C:/Users/Ashwini/Downloads/Titanic-Dataset.csv")
data.head()
```

```python
data.info()
```

```python
# The Titanic dataset contains demographic and Socio-economic informaation of pase
```

```python
sns.countplot(x='Sex', data=data)
plt.title('Gender Distribution of Passenger')
plt.show()
```

```python
sns.barplot(x='Sex',y='Survived',data=data)
plt.title("Survival Rate by Gender")
plt.show()
```

```python
sns.barplot(x='Pclass',y='Survived',data=data)
plt.title('Survival Rate by Passenger Class')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data['Age'],bins=30,kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x="Survived",y="Age",data=data)
plt.title("Age VS Survival")
plt.show()
```

```python
sns.boxplot(x="Survived",y="Fare",data=data)
plt.title("Fare VS Survival")
plt.show()
```

- Survival rates differ significantly across gender and passengers class.
- Female passenger show a higher survival rate compared to male passenger.
- Higher fare and first-class passengers tend to have higher survival probability.
- These patterns indicate potential bias in the dataset that could  influence predictive models.
